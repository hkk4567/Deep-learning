#### 2. Các thao tác trên dãy số a
(a) Bình phương các phần tử trong dãy số <br/>

(d) Tính phương sai của dãy số $\sigma^2 = \frac{\sum{\left(a_i - \mu\right)^2}}{N - 1}$ với $\mu = \frac{\sum{a_i}}{N}$ và độ lệch chuẩn $var = \sqrt{\sigma^2}$

```python
print("2. Các thao tác tính toán đơn giản với dãy số a")
""" CÁC BẠN LÀM BÀI Ở ĐÂY """
# ...
a2=np.array([])
for i in a:
  a2=np.append(a2, i**2)
na = len(a)
print("a. Binh phuong a: ", a2)
print("\nb. Do dai a: ", na)

""" CÁC BẠN LÀM BÀI Ở ĐÂY """
# ...
v_max = float("-inf")
v_min = float("inf")
v_avg = 0
for i in a:
  v_max = max(v_max,i)
  v_min = min(v_min,i)
  v_avg += i
v_avg /= na

print(f'\nc. Gia tri lon nhat: [{v_max}], gia tri nho nhat: [{v_min}], gia tri trung binh [{v_avg}]')

""" CÁC BẠN LÀM BÀI Ở ĐÂY """
# ...
v_sigma2 = 0
for i in a:
  v_sigma2 += pow((i - v_avg),2)
v_sigma2 /= (na -1)
v_var = v_sigma2 ** 0.5
print(f'\nd. Phuong sai: [{v_sigma2: .2f}] va do lech chuan: [{v_var: .2f}]')
```

```python
""" CÁC BẠN LÀM BÀI Ở ĐÂY """
# ...
a = np.array([120, 183, 160, 175, 145, 162, 190, 160, 152, 162])
a_la_chan = np.array([], dtype=bool)
a_chan = np.array([])
for i in a:
  if i %2==0:
    a_la_chan = np.append(a_la_chan, True)
    a_chan = np.append(a_chan,i)
  else:
    a_la_chan = np.append(a_la_chan, False)
n_a_chan = len(a_chan)

x_la_chan = np.array([], dtype=bool)
x_chan = np.array([])
for i in x:
  if i %2==0:
    x_la_chan = np.append(x_la_chan, True)
    x_chan = np.append(x_chan,i)
  else:
    x_la_chan = np.append(x_la_chan, False)
n_x_chan = len(x_chan)
print( "b. Cac phan tu chan trong day a, b: ")
print(f"+ Day a: {a}")
print(f"  - Loc chan: {a_la_chan}")
print(f"  - Day chua phan tu chan trong a: {a_chan}, so luong {n_a_chan} phan tu.")
print(f"+ Day x: {x}")
print(f"  - Loc chan: {x_la_chan}")
print(f"  - Day chua phan tu chan trong a: {x_chan}, so luong {n_x_chan} phan tu.")
```

```python
""" CÁC BẠN LÀM BÀI Ở ĐÂY """
# ...
v_le = np.array([])
for i in range(len(a)):
  if i%2!=0:
    v_le = np.append(v_le, i)
a_le = np.array([])
x_le = np.array([])
for i in v_le:
  a_le = np.append(a_le, a[int(i)])
  x_le = np.append(x_le, x[int(i)])
hieu_le = np.array([],dtype=int)
hieu_le2 = 0
for i in range(len(a_le)):
  hieu_le = np.append(hieu_le, a_le[i] - x_le[i])
  hieu_le2 += (a_le[i] - x_le[i])**2
kc_le = hieu_le2  ** 0.5
print( "c.Tính hiệu các phần tử ở vị trí lẻ của mảng a và x: ")
print(f"+ Vi tri le trong a va x: {v_le}")
print(f"+ Day so a: {a} va cac so o vi tri le: {a_le}")
print(f"+ Day so x: {x} va cac so o vi tri le: {x_le}")
print(f"+ Hieu cua a va x o vi tri le: {hieu_le}")
print(f"+ Khoang cach cac phan tu cua a va x o vi tri le: {kc_le: .2f}")
```

#### 4. Giải hệ phương trình tuyến tính
Giải hệ phương trình tuyến tính sau:
$$
\left\{\begin{matrix}
        4x_1  + 3x_2 - 5x_3 = 2 \\
        -2x_1 - 4x_2 + 5x_3 = 5 \\
        8x_1 + 8x_2 = -3 \\
\end{matrix}\right.
$$

```python
""" CÁC BẠN LÀM BÀI Ở ĐÂY """
# ...
A = np.array([[4,3,-5], [-2,-4,5], [8,8,0]])
y = np.array([2,5,-3])
x = np.linalg.solve(A,y)
yy = A @ x
diff = np.linalg.norm(yy - y)
print(f"+ Ma tran A: \n{A}")
print(f"+ Vector y: {y}")
print(f"+ Nghiem x: {x}")
print(f"+ Kiem tra: yy = Ax = {yy} ==> ||yy -y ||_2 = {diff: .2f}")
```

### Bài 3. Tính gần đúng số pi dùng phương pháp Monte Carlo

Vẽ một đường tròn $C$ có bán kính $r$ nội tiếp một hình vuông $C$, và đặt ngẫu nhiên $N$ chấm lên hình vuông. Tỉ lệ các chấm nằm trong hình tròn trên tổng số chấm xấp xỉ bằng diện tích của hình tròn chia cho hình vuông như sau:
$$
\frac{N_C}{N} \approx  \frac{\mathbb{S}_C}{\mathbb{S}_N}
$$
nơi mà $N_C$ là các chấm trong hình tròn, $N$ là tổng số chấm, $\mathbb{S}_C$ và $\mathbb{S}_N$ lần lượt là diện tích của hình tròn và hình vuông.
$$ \mathbb{S}_C = \pi * r^2 $$
$$ \mathbb{S}_N = (2*r)^2 = 4r^2 $$
Do đó, công thức trên trở thành:
$$
\frac{N_C}{N} \approx \frac{\pi * r^2}{4r^2} = \frac{\pi}{4}
$$
Suy ra, ta có giá trị số $\pi$ được tính như sau:
$$
\pi \approx \frac{4N_C}{N}
$$

Các bạn hãy lập trình tính sấp xỉ số $\pi$ dùng phương pháp Monte Carlo. Cho biết các sai số với $N = 100$, $N = 10000$, và $N = 1000000$. Vẽ đồ thị minh họa cho chương trình.

**Hướng dẫn**: Viết chương trình phát sinh $N$ điểm $(x_i, y_i)$ với $x_i, y_i \in [-1, -1]$. $(x_i, y_i)$ thuộc hình tròn tâm $(0, 0)$ bán kính 1 nội tiếp hình vuông có tâm tại $(0, 0)$ độ dài cạnh 2 khi và chỉ khi $x_i^2 + y_i^2 <= 1$.

```python
import matplotlib.pyplot as plt # khai báo thư viện vẽ pyplot
import math

circle = plt.Circle((0, 0), 1, color='g') # tạo đối tượng vòng tròn

fig, ax = plt.subplots(figsize=(6, 6)) # lấy figure và vùng vẽ

plt.plot(0, 0, 'o', color=(0.9, 0.9, 1.0), alpha=0.8) # vẽ điểm tại tâm (0,0)
ax.add_patch(circle) # thêm vòng tròn vào vùng vẽ

x = 0.75; y = math.sqrt(1 - (x ** 2)) # (x = 0.75, y = căn bậc 2 của 1 - x^2)
plt.arrow(0, 0, x, y)

plt.xlim(-1, 1) # giới hạn trục x
plt.ylim(-1, 1) # giới hạn trục y

plt.savefig('pi.png') # lưu đồ thị xuống tập tin
plt.show()            # hiển thị đồ thị
```

```python
import numpy as np

def calc_pi_monte_carlo(n = 100):
    pi = 0

    # gợi ý: sử dụng np.random.rand(s1, s2, ...) --> sinh ngẫu nhiên các số trong nửa đoạn [0.0, 1.0)
    """ CÁC BẠN LÀM BÀI Ở ĐÂY """
    # ...
    Nc = 0
    for i in range(n):
      x = np.random.rand() * 1
      y = np.random.rand() * 1
      if x**2 + y ** 2 <= 1:
        Nc += 1
    pi = 4*Nc/n
    return pi
# calc_pi_monte_carlo
```

```python
import math

print("epsilon(n=100): ", calc_pi_monte_carlo(n = 100) - math.pi)
print("epsilon(n=100): ", calc_pi_monte_carlo(n = 100) - math.pi)
print("epsilon(n=10000): ", calc_pi_monte_carlo(n = 10000) - math.pi)
print("epsilon(n=10000): ", calc_pi_monte_carlo(n = 10000) - math.pi)
print("epsilon(n=1000000): ", calc_pi_monte_carlo(n = 1000000) - math.pi)
print("epsilon(n=1000000): ", calc_pi_monte_carlo(n = 1000000) - math.pi)
```
