```python
x1 = [0,0,1,1]
x2 = [0,1,0,1]
yd = [0,0,0,1]

w1 = 0.3
w2 = -0.2
a = 0.2
b = 0.4

def step_func(z):
  if z >= 0:
    return 1
  else:
    return 0

for epoch in range(10):
  w1_grad = 0
  w2_grad = 0

  y_pred =[]
  for i in range(len(x1)):
    y_pred.append(step_func(w1*x1[i] + w2*x2[i] - b))
  print("====================")
  print("epoch: ", epoch)
  print("yd: ", yd)
  print("y: ", y_pred)
  print("x1", x1)
  print("x2", x2)
  print("w1, w2: ", w1,", ",w2)
  miss_class = []
  e_t = []
  for i in range(len(yd)):
    if yd[i] != y_pred[i]:
      miss_class.append((x1[i], x2[i]))
      e_t.append(yd[i] - y_pred[i])
  for i in range(len(miss_class)):
    w1_grad += a*e_t[i]*miss_class[i][0]
    w2_grad += a*e_t[i]*miss_class[i][1]
  w1 += w1_grad
  w2 += w2_grad
```
